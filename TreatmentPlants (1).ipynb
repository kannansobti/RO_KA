```python
#Create common year column from WO Year & Comm Year
#1. If both are null, the result is null
#2. If one is null, the non-null value is kept
#3. If both are present, 
#3a. If values are the same, keep it
#3b. If values are different, apply the max() logic

def create_combined_year_column(df, year_col_a, year_col_b) -> pd.DataFrame:
    
    
    df['Combined_Year_Temp'] = df[year_col_a].combine_first(df[year_col_b])
    
    both_present_mask = df[year_col_a].notna() & df[year_col_b].notna()

    same_value_mask = df[year_col_a] == df[year_col_b]

    different_value_mask = both_present_mask & ~same_value_mask
    
    latest_year = df[[year_col_a, year_col_b]].max(axis=1)

    df['Combined_Year'] = np.where(
        different_value_mask, 
        latest_year, 
        df['Combined_Year_Temp']
    )

    df.drop(columns=['Combined_Year_Temp'], inplace=True)
    
    df['Combined_Year'] = df['Combined_Year'].astype('Int64') 

    return df
```

```python
#Categorization of Schemes from NRDWP SchemeType in a new column
def categorize_scheme_type(df: pd.DataFrame) -> pd.DataFrame:
    
    
    adsorption_list = ["adsorption cwpps"]
    
    defluoridation_list = [
        "electric defluoridation plant cwpps", 
        "electric defluoridation plant(edf)", 
        "hp defluoridation plant(hpdf)", 
        "including dfu", 
        "solar electrolytic defluoridation cwpps"
    ]
    
    irp_list = [
        "htw with irp", 
        "iron removal plan", 
        "irp", 
        "m-ii with irp", 
        "m-iii with irp", 
        "tp with irp", 
        "shp with irp"
    ]
    
    ro_list = [
        "including ro", 
        "reverse osmosis (ro) cwpps", 
        "ro", 
        "dug well/drinking well", 
        "filters", 
        "tera filter"
    ]
    
    nano_tech_list = ["nano technology cwpps"]
    
    wss_list = [
        "individual water supply scheme", 
        "regional water supply", 
        "regional water supply scheme"
    ]
    
    treatment_plant_list = [
        "mini treatment plant", 
        "tp",  
        "treatment plant"
    ]
    
    
    conditions = [
        # Adsorption
        df['scheme_type_lower'].isin(adsorption_list),
        
        # Defluoridation
        df['scheme_type_lower'].isin(defluoridation_list),
        
        # IRP
        df['scheme_type_lower'].isin(irp_list),
        
        # RO
        df['scheme_type_lower'].isin(ro_list),
        
        # Nano Technology
        df['scheme_type_lower'].isin(nano_tech_list),
        
        # WSS (Requires matching BOTH scheme_type_lower AND the 'type' column)
        (df['type'] == "TreatmentPlant") & (df['scheme_type_lower'].isin(wss_list)),
        
        # Treatment Plant
        df['scheme_type_lower'].isin(treatment_plant_list)
    ]
    
    values = [
        "adsorption", 
        "defluoridation plant", 
        "irp", 
        "ro", 
        "nano technology", 
        "WSS", 
        "treatment plant"
    ]
    
    df['scheme_type_fixed'] = np.select(conditions, values, default= "")
    
    return df
```

```python
#Final categorization
def categorize_treatment_scheme(df: pd.DataFrame) -> pd.DataFrame:
    
    df['scheme_name_lower'] = df['scheme_name_lower'].astype(str)
    df['scheme_type_fixed'] = df['scheme_type_fixed'].astype(str)
    df['scheme_type_lower'] = df['scheme_type_lower'].astype(str) 
    df['type'] = df['type'].astype(str)

    conditions = []
    values = []

    conditions.append(df['scheme_type_fixed'] == "adsorption")
    values.append("Adsorption")

    conditions.append((df['scheme_type_fixed'] == "nano technology") & (df['scheme_type_fixed'] != "adsorption"))
    values.append("Nano Technology")
    
    iarp_regex = r"\b(i\.?\s*a\.?\s*r\.?\s*p\.?|a\.?\s*i\.?\s*r\.?\s*p\.?|iron (and|&)? arsenic removal( plant)?|arsenic\s+(and|&)?\s+iron(\s+removal( plant)?)?|iarp\w*|iarp|airp)\b"
    conditions.append((df['scheme_name_lower'].str.contains(iarp_regex, regex=True, case=False)) & (df['scheme_type_fixed'] != "nano technology") & (df['scheme_type_fixed'] != "adsorption"))
    values.append("IARP")

    irp_regex = r"\b(i\.?\s*r\.?\s*p\.?|iron removal( plant)?|irp\w*|mirp|iiron|iep|ir)\b"
    conditions.append(
        ((df['scheme_type_fixed'] == "irp") |
        (df['scheme_name_lower'].str.contains(irp_regex, regex=True, case=False)) & (df['scheme_type_fixed'] != "nano technology") & (df['scheme_type_fixed'] != "adsorption") & (df['scheme_type_fixed'] != "IARP") & (~(df['scheme_name_lower'].str.contains(iarp_regex, regex=True, case=False)))))
    values.append("IRP")

    arp_regex = r"\b(a\.?\s*r\.?\s*p\.?|arsenic removal( plant)?|arp\w*|arp|arsenic)\b"
    conditions.append(df['scheme_name_lower'].str.contains(arp_regex, regex=True, case=False))
    values.append("ARP")

    ro_name_regex = r"\b(r\.?\s*o\.?|reverse osmosis(\s*\(ro\))?|reverse osmosis|roplant|drinking water|drinkingwater|dwf|d\.?\s*w\.?|dw)\b"
    ro_tp_np_condition = (df['scheme_type_fixed'] == "treatment plant") & df['scheme_name_lower'].str.contains(r"\bnp\b", regex=True, case=False)
    ro_tp_lps_condition = (df['scheme_type_fixed'] == "treatment plant") & df['scheme_name_lower'].str.contains(r"\blps\b", regex=True, case=False)
    
    conditions.append(
        (df['scheme_type_fixed'] == "ro") |
        df['scheme_name_lower'].str.contains(ro_name_regex, regex=True, case=False) |
        ro_tp_np_condition |
        ro_tp_lps_condition
    )
    values.append("RO")

    dfp_regex = r"\b(d\.?\s*f\.?\s*(plant)?|f\.?\s*r\.?\s*p\.?|df plant|defluoridation( plant)?|de-?fluoridation|fluoride( plant)?|fluoridation|fluorosis|florid(e|ation)?|fluridation|flouried|fluride|fluroridei?|de fluoridetion|d floraition|d fluorisetion|dfu|edf|sdfu|frp(_\w*)?|fru|df\w*)\b"
    conditions.append(
        (df['scheme_type_fixed'] == "defluoridation plant") |
        df['scheme_name_lower'].str.contains(dfp_regex, regex=True, case=False)
    )
    values.append("Defluoridation Plant")

    purification_name_regex = r"\b(w\.?\s*p\.?\s*p\.?|purification|wpp_\w+|cwpp|cwpps|water purification plant|purificatin|wp plant|wpp_ten)\b"
    purification_condition = (df['type'] == "TreatmentPlant") & df['scheme_name_lower'].str.contains(purification_name_regex, regex=True, case=False)
    
    conditions.append(
        (df['scheme_type_lower'] == "ion exchange cwpps") |
        purification_condition
    )
    values.append("Purification")

    wss_name_regex = r"\b(w\.?\s*s\.?\s*s\.?|p\.?\s*w\.?\s*s\.?\s*s\.?|p\.?\s*w\.?\s*s\.?|mpws|piped water supply scheme|(piped)? water supply scheme|cwpps|water purification plant|purificatin|rpws|w/s/s|r\.w\.s\.s\.-m1|w/ scheme|rws|water supply|wss)\b"
    wss_tp_condition = (df['type'] == "TreatmentPlant") & df['scheme_name_lower'].str.contains(wss_name_regex, regex=True, case=False)
    
    conditions.append(
        (df['scheme_type_fixed'] == "WSS") |
        wss_tp_condition
    )
    values.append("Water Supply")

    conditions.append(df['scheme_name_lower'].str.contains(r"\bdahp\b", regex=True, case=False))
    values.append("DAHP Treatment Plant")

    others_condition = (
        (df['scheme_type_fixed'] == "treatment plant") |
        (df['type'] == "TreatmentPlant") |
        df['scheme_name_lower'].str.contains("treatment plant", regex=True, case=False)
    )
    conditions.append(others_condition)
    values.append("Others")


    
    df['treatment_scheme_type'] = np.select(conditions, values, default="")

    df = df[df['treatment_scheme_type'] != ""].copy()
    
    second_drop_mask = (
        (df['treatment_scheme_type'] == "Others") & 
        (
            (df['scheme_name_lower'] == "spot source") | 
            (df['scheme_name_lower'] == "s") |
            df['scheme_name_lower'].str.contains(r"\b(spot (source)?|awc|anganwadi|tara pump|hand pump)\b", regex=True, case=False)
        )
    )
    
    
    df = df[~second_drop_mask].copy()

    return df.reset_index(drop=True)
```

```python
#State wise Plant wise grouping
def pivot_summary_by_state(summary_df: pd.DataFrame) -> pd.DataFrame:
    
    if summary_df.empty:
        return pd.DataFrame()
        
    pivot_table = summary_df.pivot(
        index='stateid', 
        columns='treatment_scheme_type', 
        values='Unique_Plant_Count'
    ).fillna(0) 

    pivot_table = pivot_table.astype(int)
    
    return pivot_table
```

```python
from IPython.display import HTML, display
import base64
import io

def download_excel_file(df, filename="StateWise TP.xlsx", sheet_name="Sheet1", include_index=False):
    """
    Converts a pandas DataFrame to a true Excel file (.xlsx) in memory and 
    creates a downloadable link, ensuring proper column separation.

    Args:
        df (pd.DataFrame): The DataFrame to export.
        filename (str): The name for the downloaded Excel file.
        sheet_name (str): The name of the sheet within the Excel file.
        include_index (bool): Whether to include the DataFrame index as a column.
    """
    # 1. Save the DataFrame to an in-memory binary buffer
    output = io.BytesIO()
    
    # 2. Use pd.ExcelWriter to save the data in the .xlsx format
    # The 'openpyxl' engine handles the structure, guaranteeing columns are separated.
    with pd.ExcelWriter(output, engine='openpyxl') as writer:
        df.to_excel(writer, sheet_name=sheet_name, index=include_index)
    
    # Reset the buffer's pointer to the beginning to read the content
    output.seek(0)
    
    # 3. Encode the Excel file bytes for the download link
    b64 = base64.b64encode(output.read()).decode()
    
    # 4. Create the HTML download link (using the MIME type for .xlsx)
    mime_type = 'application/vnd.openxmlformats-officedocument.spreadsheetml.sheet'
    href = f'<a href="data:{mime_type};base64,{b64}" download="{filename}">✅ Click Here to Download Proper Excel File: {filename}</a>'
    
    # 5. Display the link
    display(HTML(href))
```
